```python
import matplotlib.pyplot as plt
import numpy as np

aqi = np.random.uniform(0,300,10000)
hospital = aqi * np.random.uniform(0.5,1.5,10000)

plt.hexbin(aqi,hospital,gridsize = 30 ,cmap = 'Purples')

cb = plt.colorbar(label = 'Count')

plt.xlabel('AQI')
plt.ylabel('Hospital')
plt.title('AQI and Hospital')
plt.show()
```

```python
#桑基图 （Sankey Diagram）
import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go

labels = ['coal','solar','wind','nuclear','residential','industrial','commercial']
sources = [0,1,2,3,0,1,2,3]
target = [5,5,5,5,6,6,6,6]
values = [100,200,300,400,100,200,300,400]

fig = go.Figure(data = [go.Sankey(node = dict(pad = 15,thickness = 20,line = dict(color = 'black',width = 0.5),label = labels),link = dict(source = sources,target = target,value = values))])
fig.update_layout(title_text = 'Energy Supply',font_size = 10)
fig.show()
```

## 三维曲面图
#### 三维曲面图展示三个变量的函数关系，通过颜色的变化来展示高度信息。
#### 三维曲面图中，数据点被映射到一个三维坐标系中的曲面上
#### 通过形状、颜色、高度等信息来展示数据的三维特征。

```python
import plotly.graph_objects as go
import numpy as np

x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

x,y = np.meshgrid(x,y)
z = np.sin(np.sqrt(x**2 + y**2))

fig = go.Figure(data = [go.Surface(x = x,y = y,z = z)])
fig.update_layout(title = '3D Surface Plot',scene = dict(
    xaxis_title = 'XAxis',
    yaxis_title = 'YAxis',
    zaxis_title = 'ZAxis'
))
fig.show()
```

## 雷达图
### 雷达图是一种表现多维数据的强弱的图表，它将多个维度的数据量映射到坐标轴上，这些坐标轴起始于同一个圆心点，通常结束于圆周边缘，将一组点使用线连接起来成为雷达图


```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd

categ = ['speed','reliability','comfort','safety','efficiency']
values = [90,75,88,92,85]

fig = go.Figure()
fig.add_trace(go.Scatterpolar(
    r=values,
    theta=categ,
    fill='toself',
    name='Product A'
))
fig.update_layout(title = 'R')
fig.show()
```

```python
# 三维气泡图
import plotly.graph_objects as go
import numpy as np

np.random.seed(42)
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fog = go.Figure(data = [go.Scatter3d(
    x = x,
    y = y,
    z = z,
    mode = 'markers',
    marker = dict(
        size = 12,
        color = z,
        opacity = 0.8,
        colorscale = 'Viridis'
    )
)])
fog.update_layout(
    scene = dict(
        xaxis = dict(nticks = 4, range = [0, 1],),
        yaxis = dict(nticks = 4, range = [0, 1],),
        zaxis = dict(nticks = 4, range = [0, 1],),),
    width = 700,
    margin = dict(r = 20, l = 10, b = 10, t = 10)
)
fog.show()
```

## 漏斗图
#### 漏斗图是用于业务流程比较规范，周期较长，环节多的单流程单项分析的一种图表。


```python
import plotly.graph_objects as go
values = [600, 300, 200, 100, 400, 500]
fig = go.Figure(go.Funnel(y = ['Website visit', 'Downloads', 'Potential customers', 'Requested price', 'invoice sent'],
                          x = values, textinfo = 'value+percent initial'))
fig.update_layout(title = 'Sales funnel of a deal')
fig.show()
```
